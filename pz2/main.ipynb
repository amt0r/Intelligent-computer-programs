Увесь механізм працює завдяки взаємодії трьох класів (BaseNeuron, Neuron та NeuralNetwork), які крок за кроком будують і навчають нейрону мережу. Вони беруть на себе все: від створення базових нейронів із випадковими вагами та функціями активації (Sigmoid чи ReLU), до їх об'єднання у повноцінну мережу. Фактично, ці класи дозволяють системі робити дві ключові речі — пропускати через себе дані для передбачень (прямий хід) і, що найголовніше, аналізувати свої помилки за допомогою MSE або Cross-Entropy. Розуміючи, де вона помилилася, мережа самостійно оновлює свої зв'язки (через алгоритм зворотного поширення похибки), щоб з кожною епохою ставати розумнішою і точнішою.

```python
import random
import math


class BaseNeuron:
    def __init__(self, num_inputs, weights=None, bias=None):
        if weights is not None:
            self.weights = weights
        else:
            self.weights = [random.uniform(-0.5, 0.5) for _ in range(num_inputs)]
            
        self.bias = bias if bias is not None else random.uniform(-0.5, 0.5)

    def calculate_sum(self, inputs):
        return sum(x * w for x, w in zip(inputs, self.weights)) + self.bias

class Neuron(BaseNeuron):
    def __init__(self, num_inputs, activation_type='sigmoid'):
        super().__init__(num_inputs)
        self.activation_type = activation_type
        self.last_inputs = []
        self.last_output = 0.0
        self.delta = 0.0

    def forward(self, inputs):
        self.last_inputs = inputs
        S = self.calculate_sum(inputs)
        
        if self.activation_type == 'sigmoid': 
            self.last_output = 1.0 / (1.0 + math.exp(-S))
        elif self.activation_type == 'relu':
            self.last_output = max(0.0, S)
            
        return self.last_output

    def get_derivative(self):
        if self.activation_type == 'sigmoid':
            return self.last_output * (1.0 - self.last_output)
        elif self.activation_type == 'relu':
            return 1.0 if self.last_output > 0 else 0.0

class NeuralNetwork:
    def __init__(self, layer_sizes, activation='sigmoid', name="Unnamed Model", loss_type='mse'):
        self.name = name
        self.loss_type = loss_type
        self.layers = []
        
        for i in range(1, len(layer_sizes)):
            if i == len(layer_sizes) - 1:
                layer_activation = 'sigmoid'
            else:
                layer_activation = activation
                
            layer = [Neuron(layer_sizes[i-1], activation_type=layer_activation) for _ in range(layer_sizes[i])]
            self.layers.append(layer)

    def forward(self, inputs):
        current_inputs = inputs
        for layer in self.layers:
            current_inputs = [neuron.forward(current_inputs) for neuron in layer]
        return current_inputs 

    def backward(self, expected, learning_rate, loss_type='mse'):
        for j, output_neuron in enumerate(self.layers[-1]):
            if loss_type == 'mse':
                error = expected[j] - output_neuron.last_output
                output_neuron.delta = error * output_neuron.get_derivative()
            
            elif loss_type == 'cross_entropy':
                output_neuron.delta = expected[j] - output_neuron.last_output

        for i in reversed(range(len(self.layers) - 1)):
            for j, neuron in enumerate(self.layers[i]):
                error_sum = sum(next_n.weights[j] * next_n.delta for next_n in self.layers[i + 1])
                neuron.delta = error_sum * neuron.get_derivative()

        for layer in self.layers:
            for neuron in layer:
                for j in range(len(neuron.weights)):
                    neuron.weights[j] += learning_rate * neuron.delta * neuron.last_inputs[j]
                neuron.bias += learning_rate * neuron.delta

    def train(self, X, y, learning_rate=0.1, loss_type='mse', target_error=0.005):
        epoch = 0
        
        while True:
            epoch_loss = 0.0
            
            for i in range(len(X)):
                output = self.forward(X[i])
                
                self.backward(y[i], learning_rate, loss_type)
                
                for j in range(len(y[i])):
                    epoch_loss += (y[i][j] - output[j]) ** 2
                    
            mean_loss = epoch_loss / len(X)
            epoch += 1
            
            if mean_loss <= target_error:
                break
                
            if epoch >= 100_000:
                print(f"Зупинка! Досягнуто 100.000 епох. Фінальна похибка: {mean_loss:.6f}")
                break
                
        return epoch
```

Секундомір.

```python
import time

class Stopwatch:
    def __init__(self):
        self.start_time = None
        self.elapsed_time = 0.0
        self.is_running = False

    def start(self):
        if not self.is_running:
            self.start_time = time.perf_counter()
            self.is_running = True

    def stop(self):
        if self.is_running:
            self.elapsed_time += time.perf_counter() - self.start_time
            self.is_running = False
        return self.elapsed_time

    def reset(self):
        self.start_time = None
        self.elapsed_time = 0.0
        self.is_running = False

    def get_time(self):
        if self.is_running:
            return self.elapsed_time + (time.perf_counter() - self.start_time)
        return self.elapsed_time
```

Створення всіх варіантів нерйоних мереж, підготовка навчальних даних, тренування кожної моделі, тестування.

Нейронні мережі у цьому експерименті включають у себе 20 різних конфігурацій, створених шляхом перебору всіх можливих комбінацій трьох ключових параметрів: п'яти варіантів архітектури (від найпростішої без прихованих шарів до "вузької" на 4 нейрони, "базової" на 16, "широкої" на 64 та двошарової на 16 і 8 нейронів), двох функцій активації (класичної Sigmoid та сучасної ReLU) і двох методів обчислення похибки під час навчання (MSE та Cross-Entropy).

```python
models = []

architectures = {
    "Без прихованих шарів": [36, 2],
    "Вузька": [36, 4, 2],
    "Базова": [36, 16, 2],
    "Широка": [36, 64, 2],
    "Два приховані шари": [36, 16, 8, 2]
}

activations = ['sigmoid', 'relu']
losses = ['mse', 'cross_entropy']

for arch_name, layers in architectures.items():
    for act in activations:
        for loss in losses:
            full_name = f"{arch_name} | {act} | {loss}"
            
            models.append(
                NeuralNetwork(layers, activation=act, name=full_name, loss_type=loss)
            )


X_train = []
y_train = []
for data in train_dataset.values():
    matrix = data["matrix"]
    target = data["target"]
    flat_matrix = [pixel for row in matrix for pixel in row]
    X_train.append(flat_matrix)
    y_train.append(target)

leaderboard = {}
timer = Stopwatch()

for model in models:
    print(f"\n{model.name}")
    
    timer.start() 
    
    epochs = model.train(X_train, y_train, learning_rate=0.2, loss_type=model.loss_type, target_error=0.001)
    
    timer.stop()
    train_time = timer.get_time()
    timer.reset()
    
    print(f"{epochs} епох ({train_time:.4f} сек).")
    
    correct_guesses = 0
    total_tests = len(test_dataset)
    
    for test_data in test_dataset.values():
        matrix = test_data["matrix"]
        expected_target = test_data["target"]
        
        flat_matrix = [pixel for row in matrix for pixel in row]
        prediction = model.forward(flat_matrix)
        binary_prediction = [1 if p >= 0.5 else 0 for p in prediction]
        
        if binary_prediction == expected_target:
            correct_guesses += 1
            
    accuracy = (correct_guesses / total_tests) * 100
    print(f"Тест: {correct_guesses}/{total_tests} ({accuracy:.1f}%)")
    
    leaderboard[model.name] = {"epochs": epochs, "accuracy": accuracy, "time": train_time}
```

Таблиця результатів містить інформацію про продуктивність 20 різних конфігурацій нейронних мереж, відображаючи для кожної з них тип архітектури, функцію активації, метод розрахунку похибки, а також підсумкову точність на тестовій вибірці, витрачений час у секундах та кількість пройдених епох. Дані відсортовані за принципом: найвищий пріоритет має точність моделі (від ідеальних 100% до найменшої), далі, серед тих, хто має однакову успішність, моделі ранжуються за реальним часом навчання (від найшвидших до найповільніших), і наостанок — за найменшою кількістю витрачених епох.

Аналіз рейтингу яскраво демонструє, що комбінація активації ReLU та функції втрат Cross-Entropy є абсолютною переможницею, повністю окупувавши топ таблиці. Cross-Entropy дозволяє мережі знаходити рішення набагато швидше за MSE (за лічені епохи замість тисяч), оскільки вона математично правильніше працює із задачами класифікації та сильніше штрафує модель за впевнені помилки. Також результати беззаперечно доводять, що моделі "Без прихованих шарів" провалили тестування (застрягли на 100 000 епох із низькою точністю) — це підтверджує, що вхідні матриці є лінійно нероздільними і для їх розуміння мережі критично необхідні приховані шари. Найкращий баланс показала двошарова та базова архітектури, тоді як занадто вузька мережа в поєднанні з MSE (17 місце) потрапила у проблему "мертвих нейронів" (Dying ReLU), не змігши підібрати правильні ваги.
